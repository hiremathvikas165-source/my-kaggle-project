```python
s_mode = df['Self_Employed'].mode()
s_mode = s_mode.values[0]
s_mode
```

```python
df['Self_Employed'] = df['Self_Employed'].fillna(s_mode)
df.isna().sum()
```

```python
c_mode = df['Credit_History'].mode()
c_mode = c_mode.values[0]
c_mode
```
